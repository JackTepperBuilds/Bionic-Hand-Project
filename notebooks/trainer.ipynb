```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import joblib
```

Read saved gesture landmark data and their corresponding names using pandas.
Separate and store the landmark data and gesture names into X and y variables.

```python
data = pd.read_csv('C:/Users/jatep/OneDrive/Desktop/gesture_data.csv.txt', header=None, sep=",")

# Check if the 63rd row or the gesture name row exists
column_name = 63
if column_name in data:
    X = data.iloc[:, 0:63]
    y = data.iloc[:, 63]

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

    rfc = RandomForestClassifier(n_estimators=100, random_state=42)
    rfc.fit(X_train, y_train)
    print("Data has been trained!")
```

```python
Prints every gesture the random forest classifier predicted based on the testing data
```

```python
prediction = rfc.predict(X_test)
print(prediction)
```

Prints the percent of how accurately the random forest classifier can predict 
gestures based on the testing data

```python
accuracy = accuracy_score(prediction, y_test, normalize=True)
print(f"% {accuracy*100}")
```

```python
test = joblib.dump(rfc, 'models.pkl')
print(test)
```
